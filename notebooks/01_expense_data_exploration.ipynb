# 📊 Personal Expense Tracker - Exploratory Data Analysis (EDA)
### Project: Personal Expense Tracker with Data Visualization
**Objective**: Ingest raw expense transaction records, clean and structure datetime features, perform exploratory data analysis, and uncover spending patterns.

```python
import os
import sys
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

# Ensure modules can be imported from src
sys.path.append(os.path.abspath('..'))
from src.data_processor import clean_expense_data
from src.analytics import get_summary_kpis, get_category_summary, get_monthly_trends

# Set visual style
plt.style.use('seaborn-v0_8-whitegrid')
%matplotlib inline
```

## 1. Load Raw Transaction Dataset

```python
csv_path = '../data/expenses_sample.csv'
raw_df = pd.read_csv(csv_path)
print(f"Raw records loaded: {len(raw_df)}")
raw_df.head(10)
```

## 2. Data Cleaning & Feature Engineering

```python
df = clean_expense_data(raw_df)
print("Cleaned DataFrame Shape:", df.shape)
df.info()
```

## 3. High-Level Summary KPIs

```python
kpis = get_summary_kpis(df)
for k, v in kpis.items():
    print(f"{k}: {v}")
```

## 4. Category-wise Spending Distribution

```python
cat_summary = get_category_summary(df)
display(cat_summary)

plt.figure(figsize=(10, 5))
sns.barplot(data=cat_summary, x='total_amount', y='category', palette='mako')
plt.title('Total Outflow by Expense Category (INR)')
plt.xlabel('Amount (₹)')
plt.ylabel('Category')
plt.show()
```

## 5. Monthly Trajectory and Seasonality

```python
monthly = get_monthly_trends(df)
display(monthly)

plt.figure(figsize=(10, 4))
plt.plot(monthly['month_name'], monthly['total_amount'], marker='o', linewidth=2.5, color='#1f77b4')
plt.title('Monthly Spending Trend')
plt.xlabel('Month')
plt.ylabel('Total Spend (₹)')
plt.grid(True, linestyle='--')
plt.show()
```
